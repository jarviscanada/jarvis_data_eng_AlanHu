## 4. Default rate across several variables

```python
q4_vars = ['NAME_CONTRACT_TYPE', 'NAME_INCOME_TYPE', 'OCCUPATION_TYPE', 'ORGANIZATION_TYPE']
# compute default rate for each variable in q4_vars
for var in q4_vars:
    default_rate = application_train_df.groupby(var)['TARGET'].mean().sort_values(ascending=False)
    display(default_rate)
```

From the above outputs, we're able to deduce the following:
- Unemployed applicants/applicants on maternity leave recorded the highest default rates by income type
- Low skill labourers recorded the highest default rates by occupation type

Perhaps occupation type and income type will be important features to focus on when building out our model

## 5. Create a 2x3 grid of distribution plots for: AMT_INCOME_TOTAL, AMT_CREDIT, AMT_ANNUITY, DAYS_BIRTH (convert to age), DAYS_EMPLOYED, EXT_SOURCE_2. Add median lines.

### First, we'll convert the days_birth feature to convert to years to make it more readable and easier to understand

```python
application_train_df['age'] = (-application_train_df['DAYS_BIRTH']) / 365
display(application_train_df[['age', 'DAYS_BIRTH']].head())
```

This is much easier to interpret; imagine describing your age as being 9500 days since birth as opposed to 25 years old

### Next, convert DAYS_EMPLOYED sentinel values

```python
application_train_df[application_train_df['DAYS_EMPLOYED'] == 365243]['DAYS_EMPLOYED']
```

We have 50,000+ rows worth of sentinel values for the DAYS_EMPLOYED feature, this would negatively impact our analysis if we don't handle these appropriately

```python
# replace sentinel value 365243 with NaN
application_train_df['DAYS_EMPLOYED'] = application_train_df['DAYS_EMPLOYED'].replace(365243, np.nan)
```

```python
application_train_df[application_train_df['DAYS_EMPLOYED'] == 365243]['DAYS_EMPLOYED'].head()
```

We have eliminated the sentinel values; this will make reading the plots more intuitive

```python
fig, axs = plt.subplots(nrows=2, ncols=3, figsize=(15,10), layout='constrained')
vars_to_plot = ['AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY', 'age', 'DAYS_EMPLOYED', 'EXT_SOURCE_2']
colours = ['blue', 'green', 'orange', 'purple', 'brown', 'pink']

# flatten arrays of axes so that we can iterate over them and assign plots
axes = axs.flatten()
for i, ax in enumerate(axes):
    col = vars_to_plot[i]
    df_col = application_train_df[col]
    df_col =df_col.clip(upper=df_col.quantile(0.99))
    sns.histplot(data=df_col, ax=ax, kde=True, color=colours[i])
    median = df_col.median()
    ax.axvline(median, color='red', linestyle='--')
    ax.set_title(col)
```

Let's break down the findings for each plot:
- AMT_INCOME_TOTAL: The distribution is skewed slightly to the right, with the median income being around 150,000
- AMT_CREDIT: The distribution is skewed slightly to the right, with the median amount of credit on a loan being around 500,000
- AMT_ANNUITY: The distribution is skewed slightly to the right, with the median installment/payment amount being around 25,000
- age: The distribution is fairly uniform, with the median age for an applicant being around 42-43
- DAYS_EMPLOYED: The distribution is skewed heavily to the left, with the median number of days of employment being around 1750
- EXT_SOURCE_2: The distribution is skewed to the right,with the median normalized external source credit score being between 0.5 and 0.6

## 6. Create a correlation heatmap of the top 15 numeric features most correlated with TARGET.

```python
numeric_df = application_train_df[numerical_cols].drop(columns=['TARGET'])
target_correlation_top15 = numeric_df.corrwith(application_train_df['TARGET']).sort_values(ascending=False).head(15)
display(target_correlation_top15)
```

These are the top 15 numeric features most correlated with TARGET, now we will display a correlation matrix showing the correlation between the 15 (pairwise)

```python
top15_df = application_train_df[target_correlation_top15.index]
corr_matrix = top15_df.corr()

# plot heatmap
plt.figure(figsize=(15,8))
sns.heatmap(data=corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
```

The above heatmap helps us to visualize the relationship/correlation between each of the top 15 numeric columns that were most correlated with the target variable. According to the colour scheme, "warmer" colours, such as orange/red, indicate high correlation, while "colder" colours like blue indicate low correlation. Some high correlation values to note include:
- 0.95 (REGION_RATING_CLIENT_W_CITY/REGION_RATING_CLIENT): The rating of the region where client lives is highly correlated with the same rating but taking the city into account. We can potentially drop one of these columns as there could be high multicollinearity
- 0.83 (LIVE_CITY_NOT_WORK_CITY/REG_CITY_NOT_WORK_CITY): One means that the client's contact address does not match work address, the other means that the client's permanent address does not match work address. Makes sense intuitively
- 0.86 (DEF_30_CNT_SOCIAL_CIRCLE/DEF_60_CNT_SOCIAL_CIRCLE): The number of people in the applicant's social circle who have ever been 30/60 days past due on a loan. Obviously, someone who has ever been over 60 DPD will automatically have been 30 DPD

## 7. Load bureau.csv. Count how many bureau records exist per application (SK_ID_CURR). Merge the count back to the main table as num_bureau_records. Visualize its relationship with default rate. 

```python
bureau_path = Path("../data/bureau.csv")
bureau_df = pd.read_csv(bureau_path)
```

```python
display(bureau_df.head(5))
display(bureau_df.describe())
display(bureau_df.info())
```

```python
# find the number of bureau records per application
bureau_records_by_app = pd.DataFrame(bureau_df.groupby(by=['SK_ID_CURR']).agg(count=('SK_ID_CURR', 'count')))
```

```python
display(bureau_records_by_app.head(5))
display(bureau_records_by_app.describe())
display(bureau_records_by_app.info())
```

### Merge back into application_train as "num_bureau_records"

```python
application_train_merged = application_train_df.merge(right=bureau_records_by_app, how='inner', on='SK_ID_CURR')
```

```python
display(application_train_merged.head(5))
```

```python
# rename the column
application_train_merged = application_train_merged.rename(columns={'count': 'num_bureau_records'})
display(application_train_merged.head(5))
```

```python
default_rate_by_bureau_records = application_train_merged.groupby('num_bureau_records')['TARGET'].mean()*100
display(default_rate_by_bureau_records)
```

```python
sns.scatterplot(data=default_rate_by_bureau_records.reset_index(), x='num_bureau_records', y='TARGET')
```

From the above scatterplot, we see that there is a slightly positive correlation between the number of records that the bureau has in its system for a particular application and the default rate. This would suggest that applications with more records tend to default more than those with fewer records.
