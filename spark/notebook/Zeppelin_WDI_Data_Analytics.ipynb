 

# Spark SQL vs Dataframe

`%sql` is the Spark SQL interpreter

`%spark.pyspark` is the PySpark interpreter

`%spark` is the Spark Scala interpreter

```scala
%sql

select count(1) from wdi_csv_parquet
```

```scala
%spark.pyspark

#Read Hive data to a df (this is lazy)
wdi_df = spark.sql("SELECT * from wdi_csv_parquet")
#Persist df in memory for fast futuer access
wdi_df = wdi_df.cache()

#Spark action is eager
z.show(wdi_df.count())
```

 

# Show Historical GDP for Canada

- Re-write the hive query (left cell) using PySpark df

```scala
%sql
SELECT year, IndicatorValue as GDP
FROM wdi_csv_parquet
WHERE indicatorCode = 'NY.GDP.MKTP.KD.ZG' and countryName = 'Canada'
ORDER BY year
```

```scala
%spark.pyspark
from pyspark.sql.functions import col

wdi_canada_df = wdi_df.filter(
    (col("IndicatorCode") == "NY.GDP.MKTP.KD.ZG") &
    (col("CountryName") == "Canada")
).orderBy('year')

z.show(
    wdi_canada_df.select(
        col("year").cast("int"),
        col("IndicatorValue")
    )
)


#use z.show to display df result and draw a bar chart
```

 

# Show GDP for Each County and Sort By Year

- Re-write the hive query (left cell) using PySpark df  
    - hint: you can create multiple DFs 

```scala
%sql
SELECT countryname,
       year,
       indicatorcode,
       indicatorvalue
FROM wdi_csv_parquet
WHERE indicatorcode = 'NY.GDP.MKTP.KD.ZG'
DISTRIBUTE BY countryname
SORT BY countryname, year
```

```scala
%spark.pyspark

query_df = wdi_df.filter(col("indicatorcode") == "NY.GDP.MKTP.KD.ZG").select(
        col("countryname"),
        col("year"),
        col("indicatorcode"),
        col("indicatorvalue")
    ).repartition(col("countryname")).sortWithinPartitions("countryname", "year")


z.show(query_df)
```

# Find the highest GDP for each country

- Re-write the hive query (left cell) using PySpark df


```scala
%sql

SELECT wdi_csv_parquet.indicatorvalue AS value, 
       wdi_csv_parquet.year           AS year, 
       wdi_csv_parquet.countryname    AS country 
FROM   (SELECT Max(indicatorvalue) AS ind, 
               countryname 
        FROM   wdi_csv_parquet 
        WHERE  indicatorcode = 'NY.GDP.MKTP.KD.ZG' 
               AND indicatorvalue <> 0 
        GROUP  BY countryname) t1 
       INNER JOIN wdi_csv_parquet 
               ON t1.ind = wdi_csv_parquet.indicatorvalue 
                  AND t1.countryname = wdi_csv_parquet.countryname
```

```scala
%spark.pyspark
from pyspark.sql.functions import max

# let's create the t1 aliased table first (max indicator value by country filtered by indicatorvalue and indicator value)
t1 = wdi_df.filter((col('indicatorcode') == 'NY.GDP.MKTP.KD.ZG') & (col('indicatorvalue') != 0)).groupBy(
    'countryname').agg(max('indicatorvalue').alias('ind'))
    
# now create the joined table
t1_alias = t1.alias("t1")
wdi_alias = wdi_df.alias("wdi")

tmp_df = t1_alias.join(
    wdi_alias,
    (col('t1.countryname') == col('wdi.countryname')) & 
    (col('t1.ind') == col('wdi.indicatorvalue')),
    "inner"
)

z.show(tmp_df.select(col('wdi.indicatorvalue').alias('value'), col('wdi.year').alias('year'), col('wdi.countryname').alias('country')))
```
